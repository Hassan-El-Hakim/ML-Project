```python
# Group lead times into ranges
lead_time_bins = [0, 7, 30, 90, 180, 365, float("inf")]
lead_time_labels = [
    "0–7 days",
    "8–30 days",
    "31–90 days",
    "91–180 days",
    "181–365 days",
    "366+ days"
]

lead_time_groups = pd.cut(
    df["LeadTime"],
    bins=lead_time_bins,
    labels=lead_time_labels,
    include_lowest=True
)

# Calculate the cancellation rate in each range
cancellation_by_lead_time = (
    df.groupby(lead_time_groups, observed=False)["IsCanceled"]
    .agg(["mean", "count"])
)

cancellation_by_lead_time["mean"] *= 100
cancellation_by_lead_time = cancellation_by_lead_time.rename(
    columns={"mean": "CancellationRatePercent"}
)

display(cancellation_by_lead_time)
```

```python

# Cancellation rate by deposit type
cancellation_by_deposit = (
    df.groupby("DepositType")["IsCanceled"]
    .agg(["mean", "count"])
)

cancellation_by_deposit["mean"] *= 100

cancellation_by_deposit = cancellation_by_deposit.rename(
    columns={"mean": "CancellationRatePercent"}
)

display(cancellation_by_deposit.sort_values(
    "CancellationRatePercent",
    ascending=False
))
```

```python

cancellation_by_segment = (
    df.groupby("MarketSegment")["IsCanceled"]
    .agg(["mean", "count"])
)

cancellation_by_segment["mean"] *= 100

cancellation_by_segment = cancellation_by_segment.rename(
    columns={"mean": "CancellationRatePercent"}
)

display(
    cancellation_by_segment.sort_values(
        "CancellationRatePercent",
        ascending=False
    )
)
```

```python
cancellation_by_customer = (
    df.groupby("CustomerType")["IsCanceled"]
    .agg(["mean", "count"])
)

cancellation_by_customer["mean"] *= 100

cancellation_by_customer = cancellation_by_customer.rename(
    columns={"mean": "CancellationRatePercent"}
)

display(
    cancellation_by_customer.sort_values(
        "CancellationRatePercent",
        ascending=False
    )
)
```

```python
"""Exploratory visualizations for the CSE382 City Hotel H2.csv dataset.

Install: pip install pandas numpy matplotlib
Run:     python visualize_h2.py H2.csv
Output:  plots/ (one plot per original feature plus summary plots and tables)

This is exploratory analysis only: do not use outcome/post-booking columns as
predictors without a task-specific leakage audit.
"""

from pathlib import Path
import re

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


EXPECTED_COLUMNS = [
    'IsCanceled', 'LeadTime', 'ArrivalDateYear', 'ArrivalDateMonth',
    'ArrivalDateWeekNumber', 'ArrivalDateDayOfMonth', 'StaysInWeekendNights',
    'StaysInWeekNights', 'Adults', 'Children', 'Babies', 'Meal', 'Country',
    'MarketSegment', 'DistributionChannel', 'IsRepeatedGuest',
    'PreviousCancellations', 'PreviousBookingsNotCanceled', 'ReservedRoomType',
    'AssignedRoomType', 'BookingChanges', 'DepositType', 'Agent', 'Company',
    'DaysInWaitingList', 'CustomerType', 'ADR', 'RequiredCarParkingSpaces',
    'TotalOfSpecialRequests', 'ReservationStatus', 'ReservationStatusDate',
]

CATEGORICAL = {
    'IsCanceled', 'ArrivalDateYear', 'ArrivalDateMonth', 'Meal', 'Country',
    'MarketSegment', 'DistributionChannel', 'IsRepeatedGuest',
    'ReservedRoomType', 'AssignedRoomType', 'DepositType', 'Agent', 'Company',
    'CustomerType', 'ReservationStatus',
}


def save(fig, path):
    fig.savefig(path, dpi=150, bbox_inches='tight')
    plt.close(fig)


def safe_filename(name):
    return re.sub(r'[^a-zA-Z0-9_-]', '_', name)


def feature_plot(df, col, folder, top_n=15):
    """Plot every feature, adapting chart type to the variable's meaning."""
    series = df[col]
    fig, ax = plt.subplots(figsize=(10, 5))

    if col == 'ReservationStatusDate':
        dates = pd.to_datetime(series, errors='coerce')
        counts = dates.dt.to_period('M').value_counts().sort_index()
        if len(counts):
            ax.plot(np.arange(len(counts)), counts.values, marker='.', linewidth=1.5)
            ticks = np.linspace(0, len(counts) - 1, min(12, len(counts)), dtype=int)
            ax.set_xticks(ticks, [str(counts.index[i]) for i in ticks], rotation=45, ha='right')
        ax.set_ylabel('Bookings with this status date')
        ax.set_xlabel('Month')
    elif col in CATEGORICAL or not pd.api.types.is_numeric_dtype(series):
        values = series.astype('string').fillna('Missing')
        counts = values.value_counts().head(top_n)
        # Preserve calendar order for arrival months, when available.
        if col == 'ArrivalDateMonth':
            months = ['January', 'February', 'March', 'April', 'May', 'June',
                      'July', 'August', 'September', 'October', 'November', 'December']
            counts = values.value_counts().reindex(months, fill_value=0)
        ax.bar(range(len(counts)), counts.to_numpy())
        ax.set_xticks(range(len(counts)), [str(x) for x in counts.index], rotation=45, ha='right')
        ax.set_ylabel('Number of bookings')
        if values.nunique() > top_n and col != 'ArrivalDateMonth':
            ax.set_xlabel(f'Top {top_n} categories (of {values.nunique()} unique)')
    else:
        numbers = pd.to_numeric(series, errors='coerce').dropna()
        if len(numbers):
            unique = numbers.nunique()
            if unique <= 20:
                counts = numbers.value_counts().sort_index()
                ax.bar(counts.index.astype(str), counts.values)
                ax.tick_params(axis='x', rotation=45)
            else:
                # Restrict histogram range only when extreme values would
                # make the central distribution unreadable. Do not drop data
                # from descriptive-statistics tables.
                low, high = numbers.quantile([0.01, 0.99])
                clipped = numbers[(numbers >= low) & (numbers <= high)]
                ax.hist(clipped, bins=40, edgecolor='white', linewidth=0.4)
                if len(clipped) < len(numbers):
                    ax.set_xlabel('Value (histogram shows 1st–99th percentiles)')
            ax.set_ylabel('Number of bookings')

    ax.set_title(f'{col} | Missing: {series.isna().sum():,} ({series.isna().mean():.1%})')
    ax.grid(axis='y', alpha=0.2)
    save(fig, folder / f'{safe_filename(col)}.png')


def overview_plots(df, out):
    missing = df.isna().sum().sort_values(ascending=False)
    fig, ax = plt.subplots(figsize=(11, 8))
    ax.barh(missing.index[::-1], (100 * missing / len(df)).to_numpy()[::-1])
    ax.set_xlabel('Missing values (%)')
    ax.set_title('Missing values by feature')
    ax.grid(axis='x', alpha=0.2)
    save(fig, out / 'summary_missing_values.png')

    unique = df.nunique(dropna=True).sort_values(ascending=False)
    fig, ax = plt.subplots(figsize=(11, 8))
    ax.barh(unique.index[::-1], unique.values[::-1])
    ax.set_xscale('symlog', linthresh=10)
    ax.set_xlabel('Number of unique values (symlog scale)')
    ax.set_title('Feature cardinality')
    save(fig, out / 'summary_unique_values.png')

    numeric = df.select_dtypes(include=np.number).drop(columns=['IsCanceled'], errors='ignore')
    if numeric.shape[1] >= 2:
        corr = numeric.corr(numeric_only=True)
        fig, ax = plt.subplots(figsize=(12, 10))
        im = ax.imshow(corr, vmin=-1, vmax=1, cmap='coolwarm')
        ax.set_xticks(range(len(corr)), corr.columns, rotation=90, fontsize=8)
        ax.set_yticks(range(len(corr)), corr.columns, fontsize=8)
        ax.set_title('Pearson correlation among numeric features (exploratory)')
        fig.colorbar(im, ax=ax, shrink=0.75)
        save(fig, out / 'summary_numeric_correlations.png')

    # Individual outlier boxplots (1st to 99th percentile shown for readability).
    numeric_cols = [c for c in numeric if numeric[c].nunique(dropna=True) > 10]
    if numeric_cols:
        ncols = 3
        nrows = int(np.ceil(len(numeric_cols) / ncols))
        fig, axs = plt.subplots(nrows, ncols, figsize=(15, 3.1 * nrows))
        for ax in np.asarray(axs).ravel():
            ax.set_visible(False)
        for ax, col in zip(np.asarray(axs).ravel(), numeric_cols):
            ax.set_visible(True)
            s = numeric[col].dropna()
            lower, upper = s.quantile([0.01, 0.99])
            s = s[(s >= lower) & (s <= upper)]
            if len(s):
                ax.boxplot(s.to_numpy(), vert=False)
            ax.set_title(col, fontsize=10)
        fig.suptitle('Numeric feature boxplots (1st–99th percentiles displayed)')
        fig.tight_layout()
        save(fig, out / 'summary_numeric_boxplots.png')

    if {'IsCanceled', 'ArrivalDateMonth'}.issubset(df):
        order = ['January', 'February', 'March', 'April', 'May', 'June',
                 'July', 'August', 'September', 'October', 'November', 'December']
        grouped = df.groupby('ArrivalDateMonth')['IsCanceled'].agg(['mean', 'size']).reindex(order)
        fig, ax = plt.subplots(figsize=(11, 5))
        ax.plot(grouped.index, grouped['mean'] * 100, marker='o')
        ax.set_ylabel('Cancellation rate (%)')
        ax.set_title('Cancellation rate by arrival month (descriptive only)')
        ax.tick_params(axis='x', rotation=45)
        save(fig, out / 'summary_cancellation_by_month.png')

    if {'ADR', 'IsCanceled'}.issubset(df):
        fig, ax = plt.subplots(figsize=(8, 5))
        samples = []
        for val in [0, 1]:
            s = pd.to_numeric(df.loc[df.IsCanceled == val, 'ADR'], errors='coerce').dropna()
            if len(s):
                lo, hi = s.quantile([.01, .99])
                s = s[(s >= lo) & (s <= hi)]
            samples.append(s.to_numpy())
        ax.boxplot(samples, tick_labels=['Not canceled', 'Canceled'], showfliers=False)
        ax.set_ylabel('ADR')
        ax.set_title('ADR grouped by cancellation outcome (descriptive only)')
        save(fig, out / 'summary_adr_by_cancellation.png')


# Notebook settings: change these two paths if needed.
CSV_PATH = data_path
OUTPUT_DIR = 'plots'
TOP_CATEGORIES = 15

csv = Path(CSV_PATH)
if not csv.is_file():
    raise FileNotFoundError(f'{csv} not found. Upload H2.csv to your notebook folder or update CSV_PATH.')

out = Path(OUTPUT_DIR)
out.mkdir(parents=True, exist_ok=True)
features = out / 'features'
features.mkdir(exist_ok=True)
df = pd.read_csv(csv, low_memory=False)
df.columns = df.columns.str.strip()

missing_cols = [c for c in EXPECTED_COLUMNS if c not in df.columns]
extra_cols = [c for c in df.columns if c not in EXPECTED_COLUMNS]
if missing_cols:
    print('WARNING: Expected features missing:', ', '.join(missing_cols))
if extra_cols:
    print('Additional columns:', ', '.join(extra_cols))

summary = pd.DataFrame({
    'dtype': df.dtypes.astype(str),
    'non_null': df.notna().sum(),
    'missing_count': df.isna().sum(),
    'missing_percent': (100 * df.isna().mean()).round(2),
    'unique_values': df.nunique(dropna=True),
})
summary.to_csv(out / 'feature_summary.csv')
df.describe(include='all').T.to_csv(out / 'descriptive_statistics.csv')
print(f'Rows: {len(df):,} | Columns: {df.shape[1]} | Duplicate rows: {df.duplicated().sum():,}')
for index, col in enumerate(df.columns, start=1):
    feature_plot(df, col, features, top_n=TOP_CATEGORIES)
    print(f'  [{index}/{len(df.columns)}] {col}')
overview_plots(df, out)
print(f'Done. All charts saved in {out.resolve()}')
print('Exploratory plots only: review target leakage before model training.')
summary
```

```python
# ONE NOTEBOOK CELL: IsCanceled vs every other H2 feature
# Requirements: pandas, numpy, matplotlib
from pathlib import Path
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

csv_path = data_path
if not csv_path.exists():
    raise FileNotFoundError('Place H2.csv beside your notebook (or change csv_path).')

df = pd.read_csv(csv_path)
if 'IsCanceled' not in df.columns:
    raise ValueError('The CSV must contain the original IsCanceled column.')

# Only rows with a valid binary target can contribute to cancellation rates.
df['IsCanceled'] = pd.to_numeric(df['IsCanceled'], errors='coerce')
df = df[df['IsCanceled'].isin([0, 1])].copy()

output = Path('plots_iscanceled')
output.mkdir(exist_ok=True)

categorical_columns = {
    'ArrivalDateYear', 'ArrivalDateMonth', 'Meal', 'Country',
    'MarketSegment', 'DistributionChannel', 'IsRepeatedGuest',
    'ReservedRoomType', 'AssignedRoomType', 'DepositType',
    'Agent', 'Company', 'CustomerType', 'ReservationStatus'
}
month_order = ['January', 'February', 'March', 'April', 'May', 'June',
               'July', 'August', 'September', 'October', 'November', 'December']

# Plot the target itself first.
fig, ax = plt.subplots(figsize=(6, 4))
counts = df['IsCanceled'].value_counts().reindex([0, 1], fill_value=0)
ax.bar(['Not canceled (0)', 'Canceled (1)'], counts.values)
ax.set(title='Overall booking outcomes', ylabel='Number of bookings')
fig.tight_layout()
fig.savefig(output / '00_IsCanceled_distribution.png', dpi=150)
plt.show()
plt.close(fig)

results = []
for column in df.columns:
    if column == 'IsCanceled':
        continue

    feature = df[column]
    fig, ax = plt.subplots(figsize=(10, 5))
    x_label = column

    if column == 'ReservationStatusDate':
        # Final status date is a post-outcome field (leakage risk).
        dates = pd.to_datetime(feature, errors='coerce')
        group = dates.dt.to_period('M').astype('string').fillna('Missing')
        rates = df.groupby(group, dropna=False)['IsCanceled'].agg(['mean', 'size'])
        rates = rates.drop(index='Missing', errors='ignore').sort_index()
        if len(rates):
            xs = np.arange(len(rates))
            ax.plot(xs, rates['mean'] * 100, marker='.', linewidth=1.5)
            tick_idx = np.linspace(0, len(xs)-1, min(12, len(xs)), dtype=int)
            ax.set_xticks(tick_idx, rates.index[tick_idx], rotation=45, ha='right')
        x_label = 'Reservation status date (month)'
        kind = 'date (monthly cancellation rate)'
    else:
        numeric = pd.to_numeric(feature, errors='coerce')
        # Categories include IDs and low-cardinality numeric variables.
        use_categories = (column in categorical_columns
                          or not pd.api.types.is_numeric_dtype(feature)
                          or numeric.nunique(dropna=True) <= 12)

        if use_categories:
            labels = feature.astype('string').fillna('Missing')
            # For high-cardinality categories only show the most common 12.
            top = labels.value_counts().head(12).index
            if column == 'ArrivalDateMonth':
                top = [m for m in month_order if m in set(labels)]
            selected = labels.isin(top)
            grouped = df.loc[selected].groupby(labels[selected])['IsCanceled'].agg(['mean', 'size'])
            if column == 'ArrivalDateMonth':
                grouped = grouped.reindex(top).dropna()
            else:
                grouped = grouped.sort_values('size', ascending=False)
            ax.bar(range(len(grouped)), grouped['mean'] * 100)
            ax.set_xticks(range(len(grouped)), grouped.index.astype(str), rotation=45, ha='right')
            kind = 'category cancellation rate'
        else:
            # Equal-frequency bins show how cancellation changes with feature value.
            valid = numeric.notna()
            try:
                bins = pd.qcut(numeric[valid], q=10, duplicates='drop')
            except ValueError:
                bins = pd.Series(dtype='category')
            if len(bins):
                grouped = df.loc[valid].groupby(bins, observed=True)['IsCanceled'].agg(['mean', 'size'])
                ax.bar(range(len(grouped)), grouped['mean'] * 100)
                ax.set_xticks(range(len(grouped)), [str(i) for i in grouped.index], rotation=40, ha='right')
            else:
                ax.text(0.5, 0.5, 'Not enough numeric values', ha='center', va='center', transform=ax.transAxes)
            x_label = f'{column} (quantile ranges)'
            kind = 'binned numeric cancellation rate'

    ax.set_title(f'{column} vs IsCanceled')
    ax.set_xlabel(x_label)
    ax.set_ylabel('Cancellation rate (%)')
    ax.set_ylim(0, 100)
    ax.grid(axis='y', alpha=0.2)
    fig.tight_layout()
    filename = re.sub(r'[^A-Za-z0-9_-]', '_', column)
    fig.savefig(output / f'{filename}_vs_IsCanceled.png', dpi=150, bbox_inches='tight')
    plt.show()
    plt.close(fig)
    results.append({'Feature': column, 'Visualization': kind,
                    'Missing (%)': round(feature.isna().mean() * 100, 2),
                    'Unique values': feature.nunique(dropna=True)})

summary = pd.DataFrame(results)
summary.to_csv(output / 'feature_vs_iscanceled_summary.csv', index=False)
print(f'Done: {len(results) + 1} plots saved to {output.resolve()}')
print('WARNING: ReservationStatus and ReservationStatusDate reveal final outcomes; do not use as cancellation predictors.')
display(summary)  # Jupyter notebook table
```
